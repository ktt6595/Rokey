```python
# 주어진 문자열을 공백으로 구분하여 리스트에 저장한 후, 각 단어를 인덱스와 함께 딕셔너리에 저장하는 프로그램을 작성하시오.
article = '파이썬은 1991년 네덜란드계 소프트웨어 엔지니어인 귀도 반 로섬이 발표한 고급 프로그래밍 언어로, "인터프리터를 사용하는 객체지향 언어"이자 플랫폼에 독립적인, 동적 타이핑 대화형 언어이다. '
article_list = article.split(' ')
article_dict = {}

for index, word in enumerate(article_list):
  article_dict[index] = word

article_dict
```

```python
# 연속된 문자의 개수를 세어 "aaabbc" → "a3b2c1"로 변환하세요.
#  a a  a   b   b  c
s = 'aaabbc'
# 개수셀변수
count = 1
# 결과변수
res = '' # a3b2c1

# 반복문
for i in range(1, len(s)):
  if s[i] == s[i-1]:
    count += 1
  else:
    res += s[i-1] + str(count)
    count = 1
res += s[-1] + str(count)

res
```
