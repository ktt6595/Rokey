```python
# 중첩 dict
students = {
    "student1" : {
        "name": "Alice",
        "age" : 25,
        "major" : "Computer Science"
    },
    "student2" : {
        "name" : "Bob",
        "age" : 22,
        "major" : "Mathematics"
    },
    "student3" : {
        "name" : "Charlie",
        "age" : 30,
        "major" : "Economics"
    }
}
students
```

```python
students['student1']['name']
```

```python
students['student2']['major']
```

```python
# student4 새로운 학생 추가
students['student4'] = {
    "name":"David",
    "age":29,
    "major":"Biology"
}
# student1의 값 수정 Evi, 25, Education
students.update({"student1":
  {
      "name":"Evi",
      "age":25,
      "major":"Education"
  }
})

students["student1"]["age"] = 25

# student2 삭제
del students["student2"]

students
```

```python
for k, v in students.items():
  print(k, v)
```

```python
# 기존 딕셔너리의 값을 두배로 변경
original = {'a':1, 'b':2, 'c':3}

for k, v in original.items():
  print(k, v)

doubles = {k: v*2  for k , v in original.items()}
doubles
```

```python
#   {k : v      for item in iterable(반복가능한객체)   if 조건문  }
```

```python
person = {"name": "Alice", "age": 25, "city": "New York"}
# age 값 출력
print(person['age'])
# gender : female 추가
person['gender'] = 'famle'
print(person)
# 키값만 리스트로 출력
print(list(person.keys()))
# name Evi 로 변경
person['name'] = 'Evi'
print(person)
# nums = [1,2,3,4,5] 를 이용해서 키는 숫자, 값은 숫자 *2 인 딕셔너리 생성
nums = [1,2,3,4,5]
result = { n: n*2  for n in nums  }
print(result)
```
