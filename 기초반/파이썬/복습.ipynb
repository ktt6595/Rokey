```python
# 1부터 100까지의 수 중에서 3과 5의 공배수만 출력하세요..
for i in range(1, 101):
  if i % 3 == 0 and i % 5 == 0:
    print(i)
```

```python
# 1부터 50까지의 수 중 짝수의 합과 홀수의 합을 각각 구해 출력하는 프로그램을 작성하세요.
even_sum = 0
odd_sum = 0
for i in range(1, 51):
  if i % 2 == 0:
    even_sum += i
  else:
    odd_sum += i
even_sum, odd_sum
```
