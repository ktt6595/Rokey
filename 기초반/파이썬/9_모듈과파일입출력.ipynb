```python
# log.txt 파일의 내용 중 "error"를 포함하는 모든 줄만 출력하는 코드를 작성하세요.
# if 조건문
with open('log.txt', 'r') as file:
  for line in file:
    if 'error' in line:
      print(line)
```

```python
with open("file.txt", "w") as file:
    file.write("apple pie\n")
    file.write("apple juice\n")
    file.write("I like apple\n")

# file.txt 파일에서 "apple"을 "orange"로 바꾸는 코드를 작성하세요.
# replace
with open('file.txt', 'r') as file:
  content = file.read()

content = content.replace('apple', 'orange')

with open('file.txt', 'w') as file:
  file.write(content)
```
