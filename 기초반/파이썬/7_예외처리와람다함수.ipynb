```python
# 사용자가 입력한 숫자가 0일 경우 ZeroDivisionError가 발생하지 않도록 예외 처리 코드를 작성하세요.
try:
    num = int(input("숫자를 입력하세요: "))
    result = 10 / num
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다. 다른 숫자를 입력하세요.")
except ValueError:
    print("유효하지 않은 숫자입니다. 숫자를 입력해주세요.")
else:
    print(f"결과는 {result}입니다.")
```

```python
# filter(함수, iterable)
def is_even(x):
  return x % 2 == 0

numbers = [1,2,3,4,5]
result = list(filter(is_even, numbers))
result
```

```python
evnes = list(filter(lambda x: x%2==0, numbers))
evnes
```

```python
# 두 정수 a,b를 받아 그 합을 반환하는 람다 표현식 작성
a = 3
b = 5
result = lambda a, b: a+b
result(a,b)
```

```python
# 두 정수 a,b를 받아 더 큰 수를 반환하는 람다 표현식
result = lambda a, b: a if a > b else b
result(a,b)
```

```python
# 대문자로 시작하는 단어만 추출   filter
words = ["Apple", "banana", "Cat", "dog"]
result = list(filter(lambda x: x[0].isupper(), words))
result
```

```python
# 문자열 리스트에서 글자수가 5개 이상인 단어만 추출 filter
result = list(filter(lambda x: len(x) >=5, words))
result
```

```python
# reduce(함수, iterable)
from functools import reduce

nums = [3,7,2,9,5]
result = reduce(lambda x, y: x if x > y else y, nums)
result
```

```python
def multiply(x, y):
  return x*y

nums = [1,2,3,4,5]
result = reduce(multiply, nums) # 모든 요소를 곱합
result
```

```python
result = reduce(lambda x, y: x*y, nums)
result
```
