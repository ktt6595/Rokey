```python
'''
가변인자
*args
'''
def add_nums(*args):
  total = sum(args)
  print(total)

add_nums(1,2,3,4,5)
```

```python
def greet(msg, *names):
  print(names)
  for n in names:
    print(f'{msg}, {n}!')
greet('hello','rokey','boot','camp')
```

```python
'''
**kwargs
가변 키워드 인자
'''
def print_info(**kwargs):
  for k, v in kwargs.items():
    print(f"{k} and {v}")

print_info(name='철수', age=20, city='서울')
```

```python
def introduce(name, **details):
  print(name)
  for k, v in details.items():
    print(f"{k} and {v}")

introduce('rokey', age=25, hobby='독서', job='개발자')
```

```python
def add(a,b,c):
  return a + b + c

nums = [1,2,3]
add(*nums) # * 언패킹
```

```python
def person_info(name, age, city):
  print(f"name {name}, age {age}, city {city}")

info = {'name':'rokey', 'age':25, 'city':'seoul'}
person_info(**info) # ** 딕셔너리
```

```python
# *, **
def mix_func(a, b, c, x=10, y=20):
  print(a, b, c, x, y)

nums = (1,2,3)
my_dict = {'x':10, 'y':20}
mix_func(*nums, **my_dict)
```

```python
def show_info(title, *args, **kwargs):
  print(f"{title}")
  print(f"참여자 {args}")
  print(f"세부 정보 {kwargs}")
```

```python
'''
재귀 호출
'''
# 팩토리얼 5! = 5*4*3*2*1

def factorial(n): # 5 4 3 2 1
  if n == 0:
    return 1
  else:
    return n * factorial(n-1) # 5 * 4 * 3 * 2 * 1

factorial(5)
```

```python
# 문자열을 인자로 받아 거꾸로 출력하는 함수
def reversed_strings(s):
  return s[::-1]

print(reversed_strings('hello'))
```

```python
# 리스트의 평균을 구하는 함수
def average(my_list):
  return sum(my_list) / len(my_list)

average([10,20,30,40,50])
```

```python
# 가변인자 사용해서 합 구하기
def test(*args):
  return sum(args)

test(1,2,3,4,5)
```

```python
# 리스트에서 특정 값 제거하는 함수
def remove_value(my_list, value): # my_list, value
  return [  x    for x in my_list  if x != value   ]

remove_value([1,2,3,4,5,6,7], 2)
```

```python
# 가변 키워드 인자 사용
def print_info(**kwargs):
  for key, value in kwargs.items():
    print(key, value)

print_info(name='rokey', age=25)
```

```python
# 타입 힌트
def add(a:int, b:int) -> int:
  return a+b

print(add(3,7))
```

```python
def greet(name: str) -> str:
  return f'Hello {name}'

def get_first(items: list[str]) -> str:
  return items[0]

def count_scores(scores: dict[str, int]) -> int:
  return len(scores)

print(greet('rokey'))
print(get_first(["apple",'banana']))
print(count_scores({'a':90, 'b':80}))
```

```python
def square(x: int | float) -> int | float:
  return x * x

square(5), square(3.5)
```

```python
def find_value(data: dict[str, int], key: str) -> int | None:
  return data.get(key)

scores = {'rokey':100, 'bob':80}
print(find_value(scores, 'rokey'))
print(find_value(scores, 'camp'))
```

```python
def get_name(name : str | None) -> str:
  return name or 'Unknown'

print(get_name('rokey'))
print(get_name(None))
print(get_name(2))
print(get_name(True))
```

```python
from typing import Any

def first_element(items : list[Any]) -> Any:
  return items[0]

first_element([1,2,3])
```
