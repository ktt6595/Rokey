### fit / transform / fit_transform 메서드의 차이점 및 사용 시점
- .fit() ->	훈련 데이터에서 최소값/최대값 계산 / train 데이터에서만!
- .transform() -> 계산된 min/max 기준으로 변환 / test, validation 데이터
- .fit_transform() -> fit + transform을 한 번에 / train 데이터에만 사용

```테스트 데이터 정보를 미리 학습(min, max 포함)하면 모델이 미래 정보를 미리 안 것처럼 되어 버림 → 성능이 부정확해짐 ```

### MinMaxScaler란?
- MinMaxScaler는 scikit-learn(sklearn) 의 데이터 전처리 도구 중 하나로, 각 수치형(feature) 데이터를 0과 1 사이 범위로 맞춰주는(정규화, normalization) 도구
- 구조: x' = x - x(min)/x(max)-x(min), x는 raw data 값, x'는 스케일링 된 값(0~1 사이의 값)
- 해당 컬럼의 feature값의 최대값이 200, 최소값이 0일 경우 x값이 100이라면 (100-0)/(200-0) = 0.5

#### 그러면 왜 why? 이렇게 해야하는가..
- 많은 머신러닝 모델은 특징(feature) 단위가 다르면 학습이 불안정 -> 값이 다르고 단위가 다르고 그러면 범위가 맞지 않음..
- -> mileage는 km, tax는 pound, enginesize는 L 등... 이렇게 수치가 널뛰면 범위가 큰 특징이 모델에 과도한 영향을 주게 되므로

````python
# num_col minmaxscaler 해보기
from sklearn.preprocessing import MinMaxScaler

mms = MinMaxScaler()
x_train[num_col] = mms.fit_transform(x_train[num_col])
x_test[num_col] = mms.transform(x_test[num_col])

x_train.head(1)


# 훈련 데이터로만 fit → 같은 스케일 기준으로 테스트에 transform -> 이렇게 해야 데이터 누수(leakage) 방지
````

````python
from sklearn.preprocessing import LabelEncoder

for col in str_col:
  le = LabelEncoder()
  x_train[col] = le.fit_transform(x_train[col])
  x_test[col] = le.transform(x_test[col])

x_train.head()

# 문자열 → 정수로 변환(모델이 입력을 숫자로만 받기 때문)
# 각 컬럼이 0,1,2,… 정수 라벨로 치환된 x_train/x_test
# 새 라벨 인코더를 컬럼마다 새로 만들고, 훈련으로 fit 후 테스트엔 transform만 → OK
# 선형회귀 등에서 라벨 인코딩은 ‘서열’을 암시 해서 왜곡될 수 있음 → 범주 수가 많지 않다면 One-Hot을 권장
````

````python
'''
지도 학습
레이블(정답) 주어진 데이터
모델 입력과 출력간의 관계 학습
1) 분류 카테고리 예시) 이메일(스팸or정상), 손글씨(0~9사이숫자), iris(붓꽃품종)
2) 회귀 수치 예시)주식가격, 온도, 주택가격
- 단순선형회귀 X 공부시간 y 점수
X(독립변수, 입력), y(출력, 종속변수)
- 다중선형회귀
X 방갯수, 집평수, 화장실개수, 위치, 역과의거리, 역세권
y 주택가격
- 다항회귀 polynominal regression
'''
````

````python
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()
lr_model.fit(x_train, y)
lr_model.predict(x_test)
lr_model.score(x_test, y_test['price'])

# 간단한 베이스라인의 선형회귀 모델 (한 줄의 기준선 만들기)
# 입력: 스케일링 완료된 x_train, 타깃 y

# fit은 학습(가중치 계산), 반환값 없음
# predict는 예측값 반환
# score는 R²(결정계수) 반환 -> 1에 가까울수록 좋고, 음수가 나올 수도 있음(형편없는 결과라는 뜻..)

# 위에서 말씀드린 것과 같이 라벨 인코더로 해서,, 결과 수치가 낮음ㅠㅜㅜ
````
