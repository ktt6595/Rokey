# 넘파이(Numpy)
- 파이썬에서 사용되는 과학 및 수학 연산을 위한 라이브러리(Numerical Python)
- 주로 다차원 배열을 다루는 데에 특화되어 있어, 데이터 분석, 머신러닝, 과학 계산 등 다양한 분야에서 널리 사용
- 넘파이 배열은 큰 데이터셋에서 수치 연산을 수행할 때 뛰어난 성능을 보이며, 메모리 사용을 최적화하고 효율적으로 관리할 수 있음

### Numpy의 개요

1. 다차원 배열 객체 (ndarray): NumPy의 핵심은 다차원 배열을 효율적으로 다루기 위한 ndarray 객체 -> 이는 수학적으로 수학 행렬이나 텐서를 나타낼 수 있으며, 
                             숫자 데이터의 대규모 집합을 효과적으로 저장하고 조작을 가능하게 함

2. 다양한 수학 함수: NumPy는 배열 간의 산술 연산, 통계적 계산, 선형 대수학, 푸리에 변환 등 다양한 수학 함수를 제공 -> 이를 통해 복잡한 수학 연산을 간단하게 수행할 수 있음

3. 고성능: NumPy는 C 언어로 구현되어 있어, 일반적인 파이썬 코드보다 빠른 연산 속도를 제공 -> 이는 대규모 데이터 처리가 필요한 과학 계산에서 매우 중요한 포인트

4. 브로드캐스팅: NumPy는 서로 다른 크기의 배열 간의 연산을 자동으로 처리하는 브로드캐스팅 기능을 지원 -> 이를 통해 코드가 보다 간결하고 직관적으로 작성될 수 있으며,
               
5. 다양한 데이터 처리 기능: 배열 조작, 정렬, 색인, 슬라이싱 등의 다양한 데이터 처리 기능을 제공 

```python
import numpy as np
```

### 넘파이의 주요 특징과 기능




* 다차원 배열(N-dimensional array)
    - 넘파이의 핵심은 다차원 배열 ndarray
    - ndarray는 동일한 자료형을 가지는 원소들로 구성

```python
list1 = [1,2,3,4] #1차원
list2 = [[1,2,3,4,], [5,6,7,8]] #2차원
print(list1)
print(list2)
print(type(list1))
print(type(list2))
print(type(list1[0]))
print(type(list2[0]))
```

```python
ndarr1 = np.array([1,2,3,4])
print(ndarr1)
print(type(ndarr1)) # <class 'numpy.ndarray'>
print(type(ndarr1[0])) # <class 'numpy.int'>, 스칼라(scalar) 값

# 1차원 배열(vector)의 shape 확인
print(ndarr1.shape) # (4,)
```

```python
ndarr2 = np.array([[1,2,3], [4,5,6]])
print(ndarr2)
print(type(ndarr2))
print(type(ndarr2[0])) # <class 'numpy.ndarray'>, 1차원 배열(행)

# 2차원 배열(matrix)의 shape 확인
print(ndarr2.shape) # (2, 3), 2x3 행렬
```

```python
# 리스트와 ndarray 변환

# 리스트를 ndarray로 변환, ndarray는 원소사이에 ,(콤마)가 없네
list1 = [1,2,3,4]
ndarr1 = np.array(list1)
print(ndarr1)
print(type(ndarr1))
```

```python
# ndarray를 리스트로 변환
list2 = ndarr1.tolist()
print(list2)
print(type(list2))
```

#### Boolean Indexing
- Boolean값으로 이루어진 배열을 사용하여, Boolean의 True값 조건을 충족하는 원소만 선택하는 방법

#### 행렬 연산
- 넘파이에서는 다차원 배열인 ndarray를 사용하여 행렬 연산을 수행
- 행렬 연산은 선형 대수와 관련이 깊어 데이터 과학, 머신러닝, 통계 등 다양한 분야에서 사용

```python
ndarr1 = np.array([[1,2,3], [2,3,4]])
ndarr2 = np.array([[3,4,5], [1,2,3]])
print(ndarr1.shape) # .shape속성은 행렬 표기
print(ndarr2.shape)
```

```python
# 행렬 덧셈: 원리는 1행1열+1행1열, 1행2열+1행2열, ...
# 1:1 대응
print(ndarr1 + ndarr2)
```

```python
# 행렬 뺄셈
print(ndarr1 - ndarr2)
```

```python
# 행렬 원소별 곱셈 / 행렬곱(dot product)와는 다름
# 1:1 대응 곱셈
print(ndarr1 * ndarr2)
```

```python
# 행렬곱(Dot Product): =내적 연산, 벡터계산, 벡터각도, 두 벡터의 유사성 계산
# 행렬곱의 조건 맞닿는 shape가 같아야함. 떨어져 있는 shape가 결과 행렬 됨
# 예) (2,3) @ (3,2) = (2,2) -> 앞의 열과 뒤의 행이 같아야함.
# 예) (3,3) @ (3,2) = (3,2)   (2,2)@(3,2) 불가
print(ndarr1 @ ndarr2)
# ValueError: matmul: Input operand 1 has a mismatch


# ndarr3 = np.array([[1,2,3], [1,2,3], [2,3,4]])
# ndarr4 = np.array([[1,2], [3,4], [5,6]])
# print(ndarr3.shape) # (3, 3)
# print(ndarr4.shape) # (3, 2)  
```

```python
# (3,3) @ (3,2) = (3,2)
# 1행*1열=1행1열, 1행*2열=1행2열, ... 3행*2열=3행2열
# 22, 28, ..., 40
print(ndarr3 @ ndarr4)
```

```python
print(np.dot(ndarr3, ndarr4))
```

```python
# 전치 행렬(transform): 행 / 열 뒤집기, 행->열 / 열->행
print(ndarr1)
print(ndarr1.T)
```

```python
# 역행렬(reverse): 정사각 행렬에 대한 곱셈 연산으로 단위 행렬을 얻을 수 있는 행렬
# 단위 행렬: 정방행렬에 대각선이 1인 행렬
arr = np.array([[1,2], [3,4]])
print(np.linalg.inv(arr))
```

```python
# 순차적인 값 생성

arr1 = range(1,11)
print(arr1)

for i in arr1:
    print(i, end='')
```

```python
# range와 달리 np.arrange로 출력시 ndarray형태로 반환
arr1 = np.arange(1,11)
print(arr1)

for i in arr1:
    print(i, end='')
```

```python
# 정렬

ndarr1 = np.array([1,10,5,7,2,4,3,6,8,9])
print(ndarr1)
```

```python
print(np.sort(ndarr1)) #오름차순 정렬
```

```python
print(ndarr1)
```

```python
print(np.sort(ndarr1)[::-1]) # 내림 차순 정렬
```

```python
ndarr2d = np.array([[11,10,12,9], [3,1,4,2], [5,6,7,8]])
print(ndarr2d.shape)
```

```python
print(np.sort(ndarr2d, axis = 0)) #행정렬: 1열 기준으로 행순서 정렬
```

```python
print(np.sort(ndarr2d, axis = 1)) #열정렬: 기존행렬에서 행안의 열만 정렬
print(np.sort(ndarr2d, axis=1)[::-1]) #열정렬 내림차순
```

```python
print(np.sort(ndarr2d, axis=-1)) # axis는 축의 마지막 방향을 뜻함
```

# Pandas

- 데이터 분석을 위한 파이썬 라이브러리 중 하나로, 표 형태의 데이터나 다양한 형태의 데이터를 쉽게 처리하고 분석
- 특히 테이블 형식의 데이터를 다루는 데 강력한 기능을 제공하며, 데이터 프레임(DataFrame)과 시리즈(Series)라는 두 가지 주요 데이터 구조를 기반으로 동작


[주요 특징]

- 데이터 구조

    - Series: 1차원 배열과 유사한 구조로, 인덱스를 포함한 데이터를 저장

    - DataFrame: 2차원 표 형식의 데이터 구조로, 각 열이 Series 객체로 구성 -> 행과 열 모두 인덱스를 가질 수 있음



- 데이터 조작 기능

    - 데이터 선택 및 필터링: 데이터 프레임에서 특정 행이나 열을 선택하고, 조건을 기반으로 데이터를 필터링할 수 있음

    - 데이터 정렬: 행이나 열을 기준으로 데이터 정렬 가능

    - 데이터 병합 및 결합: 여러 데이터 프레임을 하나로 병합하거나 결합 가능



- 데이터 변환

    - 결측값 처리: 결측값을 쉽게 확인하고, 제거하거나 대체할 수 있음

    - 데이터 형식 변환: 데이터 형식을 다른 형식으로 변환 가능

    - 집계 및 그룹화: 데이터를 그룹별로 집계하거나 요약할 수 있음



- 입출력 기능
 
    - 다양한 파일 형식(CSV, Excel, SQL 데이터베이스, JSON 등)에서 데이터를 읽고 쓸 수 있습니다.

#### Series와 DataFrame

- Series

    - Series는 다양한 데이터 타입을 가질 수 있으며, 정수, 실수, 문자열 등 다양한 형태의 데이터를 담을 수 있음

```python
# 월별 매출 구하기
def extract_month(date):
    month =str(date.month) # 12
    if date.month < 10:
        month='0'+month # 예) 2월일 경우 02
    return str(date.year) + month # 201112.201101
```

```python
rev_by_month = retail.set_index('InvoiceDate').groupby(extract_month)['CheckoutPrice'].sum()
rev_by_month
```

```python
# xticlabels() : X축 눈금+라벨 생성
# 막대그래프 생성 함수
def plot_bar(df, xlabel, ylabel, title, titlesize=15, fontsize=12, rotation=45, figsize=(12, 6)):
    plot = df.plot(kind='bar', figsize=figsize)
    plot.set_xlabel(xlabel, fontsize=fontsize)
    plot.set_ylabel(ylabel, fontsize=fontsize)
    plot.set_title(title, fontsize=titlesize)
    plot.set_xticklabels(labels=df.index, rotation=rotation)

plot_bar(rev_by_month, xlabel='Month', ylabel='Revenue', title='Revenue By Month')
```

```python
# 요일별 매출 구하기
def extract_dow(date):
    return date.dayofweek
```

```python
# rev_by_dow = retail.set_index('InvoiceDate').groupby(extract_dow)['CheckoutPrice'].sum()
rev_by_dow = retail.set_index('InvoiceDate').groupby(lambda date: date.dayofweek)['CheckoutPrice'].sum()
rev_by_dow
```

```python
DAY_OF_WEEK = np.array(['Mon','Tue','Web','Thur','Fri','Sat','Sun'])
rev_by_dow.index = DAY_OF_WEEK[rev_by_dow.index]
rev_by_dow.index
```

```python
rev_by_dow
```

```python
plot_bar(rev_by_dow, 'DOW', 'Revenue', 'Revenue By DOW')
```

```python
# 시간대별 매출 구하기
rev_by_hour = retail.set_index('InvoiceDate').groupby(lambda date: date.hour)['CheckoutPrice'].sum()
rev_by_hour
```

```python
plot_bar(rev_by_hour, 'Hour', 'Revenue', 'Revenue By Hour')
```
