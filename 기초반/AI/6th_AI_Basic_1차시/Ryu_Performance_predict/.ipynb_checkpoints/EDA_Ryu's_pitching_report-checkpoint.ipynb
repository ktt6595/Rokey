## 류현진 데이터 로드

```python
# 류현진 데이터 로드
pitches = pd.read_csv("ryu.csv",parse_dates = True)
pitches.tail(10)
```

***********
# 데이터 전처리

## 구종 필터링 (one-hot encoding)

```python
# 구종 one_hot_encoding
one_hot = pd.get_dummies(pitches['pitch_type'])
pitches = pitches.join(one_hot)

filter_CH = pitches[pitches['CH'] == 1] # 체인지업
filter_CU = pitches[pitches['CU'] == 1] # 커브
filter_FC = pitches[pitches['FC'] == 1] # 커터
filter_FF = pitches[pitches['FF'] == 1] # 포심패스트볼
filter_FT = pitches[pitches['FT'] == 1] # 투심패스트볼

filter_CU.tail()
```

************************


## 류현진 데이터 연도별 필터링

- 경기일자 데이터에서 year만 추출하여 DataFrame에 새로운 컬럼 생성
- 연도별 원핫인코딩

```python
year = []

for idx in range(len(pitches)):

    if pitches['game_date'][idx][0:4] == '2019' :
        year.append("2019")
    elif pitches['game_date'][idx][0:4] == '2018' :
        year.append("2018")
    elif pitches['game_date'][idx][0:4] == '2017' :
        year.append("2017")
    elif pitches['game_date'][idx][0:4] == '2016' :
        year.append("2016")
    elif pitches['game_date'][idx][0:4] == '2015' :
        year.append("2015")
    elif pitches['game_date'][idx][0:4] == '2014' :
        year.append("2014")
    elif pitches['game_date'][idx][0:4] == '2013' :
        year.append("2013")
mlb_year = pd.DataFrame(year)
pitches.insert(0,"mlb_year",mlb_year)
pitches.tail()
```

```python
one_hot2 = pd.get_dummies(year)
one_hot2
pitches = pitches.join(one_hot2)
pitches.tail()
```

```python
# 원핫인코딩으로 필터링 한 데이터를 객체로 저장

filter_2019 = pitches[pitches['2019']==1]
filter_2018 = pitches[pitches['2018']==1]
filter_2017 = pitches[pitches['2017']==1]
filter_2016 = pitches[pitches['2016']==1]
filter_2014 = pitches[pitches['2014']==1]
filter_2013 = pitches[pitches['2013']==1]
```

# 데이터 분석

## 가설 1 : 류현진의 구속이 빨라졌다.

- 가설 검증절차
   1. release_speed 컬럼을 데이터의 종속변수로 설정한다.
   2. 연도별 filtering한 데이터들을 종합한 데이터프레임을 만든다.
   3. violinplot으로 dataframe을 plotting 해본다.

```python
df_release_speed = pd.DataFrame({
                  "data": [filter_2013['release_speed'].values,
                   filter_2014['release_speed'].values,
                   filter_2016['release_speed'].values,
                   filter_2017['release_speed'].values,
                   filter_2018['release_speed'].values,
                   filter_2019['release_speed'].values,
                ]
                  })



ax = sns.violinplot(data= df_release_speed['data'])
plt.title("Ryu's release speed")
plt.xlabel("year")
plt.ylabel("speed (miles)")
ax.set_xticklabels(["2013","2014","2016","2017","2018","2019"])
```

### +) catplot으로 연도별에 따른 구종별 구속분포 확인

```python
sns.catplot(x="mlb_year", y = "release_speed", hue="pitch_type", data= pitches, kind="violin",height = 9, aspect = 4)
```

```python
# 구종별 count 확인
pitches.groupby('pitch_type').size()
```

### 류현진의 동일한 구종으로 데뷔년도(2013)와의 기술통계 비교 
- 기술통계량 확인

```python
# 2013년 포심패스트볼 기술통계량
filter_2013_FF = (pitches['pitch_type'] == 'FF') & (pitches['game_date'].str.startswith("2013"))
FF_2013 = pitches[filter_2013_FF]['release_speed']
print(FF_2013.describe())
print('\n')

# 2019년 포심패스트볼 기술통계량
filter_2019_FF = (pitches['pitch_type'] == 'FF') & (pitches['game_date'].str.startswith("2019"))
FF_2019 = pitches[filter_2019_FF]['release_speed']
print(FF_2019.describe())
```

```python
# 2013년 체인지업 기술통계량
filter_2013_CH = (pitches['pitch_type'] == 'CH') & (pitches['game_date'].str.startswith("2013"))
CH_2013 = pitches[filter_2013_CH]['release_speed']
print(CH_2013.describe())
print('\n')

# 2019년 체인지업 기술통계량
filter_2019_CH = (pitches['pitch_type'] == 'CH') & (pitches['game_date'].str.startswith("2019"))
CH_2019 = pitches[filter_2019_CH]['release_speed']
print(CH_2019.describe())
```

```python
# 2013년 커브 기술통계량
filter_2013_CU = (pitches['pitch_type'] == 'CU') & (pitches['game_date'].str.startswith("2013"))
CU_2013 = pitches[filter_2013_CU]['release_speed']
print(CU_2013.describe())
print('\n')

# 2019년 커브 기술통계량
filter_2019_CU = (pitches['pitch_type'] == 'CU') & (pitches['game_date'].str.startswith("2019"))
CU_2019 = pitches[filter_2019_CU]['release_speed']
print(CU_2019.describe())
```

## 가설 1 결론
- 2019년 포심패스트볼(직구)의 평균구속은 2013년에 비해 0.36 마일 감소했고, 최고구속은 1.4 마일 감소했다.

- 2019년 체인지업의 평균구속은 2013년에 비해 0.5 마일 감소했고, 최고구속은 0.9 마일 감소했다.

### 류현진 최고구속은 오히려 감소했다. 성적향상의 원인이 아니다.

## distplot을 이용한 histogram과 pdf plotting

```python
# release extension
sns.distplot(filter_2017['release_extension'].values, bins=50, color="green", label= "2017")
sns.distplot(filter_2019['release_extension'].values, bins=50, color="darkblue", label= "2019")
plt.legend()
```

## sklearn preprocessing을 사용한 표준화 객체 만들기

```python
from sklearn import preprocessing
ex_scaled_2017 = preprocessing.scale(filter_2017['release_extension'].values)
ex_scaled_2019 = preprocessing.scale(filter_2019['release_extension'].values)


sns.distplot(ex_scaled_2017, bins=50, color="green",label= "2017")
sns.distplot(ex_scaled_2019, bins=50, color="darkblue",label= "2019")
plt.legend()
```

```python
# 2017년, 2019년 release_extension 기술통계량
print(filter_2017['release_extension'].describe())
print('\n')
print(filter_2019['release_extension'].describe())
```

```python
# release height
from sklearn import preprocessing
rz_scaled_2017 = preprocessing.scale(filter_2017['release_pos_z'].values)
rz_scaled_2019 = preprocessing.scale(filter_2019['release_pos_z'].values)


sns.distplot(rz_scaled_2017, bins=50, color="green")
sns.distplot(rz_scaled_2019, bins=50, color="darkblue")

# 2017년, 2019년 release_heihgt 기술통계량
print(filter_2017['release_pos_z'].describe())
print('\n')
print(filter_2019['release_pos_z'].describe())
```

***************
## 가설 2 결론
- 2019년 release_extension 의 std가 2017년에 비해 0.012 줄어들었다.

- 2019년 release_pos_z 의 std가 2017년에 비해 1.8 줄어들었다.

### 류현진 release point는 과거에 비해 매우 일정해졌다  -> 컨트롤이 좋아졌다.

# 총 결론

### 1. 구속이 빨라졌다.  -> 전반적으로 구속이 느려졌다.  
### 2. 볼 컨트롤이 좋아졌다. ->  release extension, release height 모두 표준편차가 줄어 좋아졌다.
### 3. 볼 무브먼트가 좋아졌다. -> release spin rate이 증가했다. (특히 커브볼이 눈에 띄게 좋아졌다)
